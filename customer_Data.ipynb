```python
# convert categorical data into numerical data
from sklearn.preprocessing import OrdinalEncoder
encoder=OrdinalEncoder()
data['Education']=encoder.fit_transform(data[['Education']])
data.head()
```

```python
#convert categorical column into numerical column using one hot encoding
data1=pd.get_dummies(data,columns=['Marital_Status'])
data1.head()
```

```python
#convert column into numerical column using one hot encoding
from sklearn.preprocessing import OneHotEncoder
Ohe=OneHotEncoder(drop='first',sparse=False)
data1=Ohe.fit_transform(data[['Marital_Status']])
data1=pd.DataFrame(data1,columns=Ohe.get_feature_names_out(['Marital_Status']))
data1.head()


```

```python
data=pd.concat([data,data1],axis=1)
data.head()
```

```python
data['Income'].fillna(data['Income'].mean(),inplace=True)
```

```python
data['Total children']=data['Kidhome']+data['Teenhome']
data['Accepted compaigns']=data['AcceptedCmp1']+data['AcceptedCmp2']+data['AcceptedCmp3']+data['AcceptedCmp4']+data['AcceptedCmp5']
data['Total deal Purchases']=data['NumDealsPurchases']+data['NumWebPurchases']+data['NumCatalogPurchases']+data['NumStorePurchases']+data['NumWebVisitsMonth']
data['Total Spend']=data['MntWines']+data['MntFruits']+data['MntMeatProducts']+data['MntFishProducts']+data['MntSweetProducts']+data['MntGoldProds']
#plot the spend on various product
spending_column=['MntWines','MntFruits','MntMeatProducts','MntFishProducts','MntSweetProducts','MntGoldProds']
for col in spending_column:
  sns.kdeplot(data[col],label=col,fill=True)
plt.legend()
plt.show()

```

**now to treat outliers**


```python
import numpy as np
from scipy.stats import zscore
from scipy.stats import norm
numerical_column=['Income','Age','Total children','Total Spend','Accepted compaigns','Total deal Purchases']
for col in numerical_column:
  z_score=np.abs(zscore(data[col]))
  data[col]=np.where(z_score>3,data[col].median(),data[col])
  outliers=np.where(z_score>3)
  data.drop(outliers[0],inplace=True)
data.head()

```

```python
#to Drop the unnecessary column
data.drop(["ID","Year_Birth","Kidhome","Teenhome",'Marital_Status',"Dt_Customer","MntWines","MntFruits","MntFishProducts","MntMeatProducts","MntSweetProducts","MntGoldProds","NumDealsPurchases","NumWebPurchases","NumCatalogPurchases","NumStorePurchases","NumWebVisitsMonth","AcceptedCmp3","AcceptedCmp4","AcceptedCmp5","AcceptedCmp2","AcceptedCmp1","Z_CostContact","Z_Revenue"],axis=1,inplace=True)
data.head()
```

```python
#feature scaling
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
data[numerical_column]=scaler.fit_transform(data[numerical_column])
data.head()
```

```python
# implementing PCA

from sklearn.decomposition import PCA
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
PCA_data=data.copy()
scaled_data=StandardScaler().fit_transform(PCA_data)
pca=PCA(n_components=2)
principalComponents=pca.fit_transform(scaled_data)
principalDf=pd.DataFrame(data=principalComponents,columns=['pC1','pC2'])
print(principalDf)
print(f'Explained variance ratios: {pca.explained_variance_ratio_}')



```

```python
plt.figure(figsize=(8, 6))
sns.pairplot(principalDf)
plt.show()
```

Applying the algorithms


```python
#plot Elbow graph to choose the number of cluster
from sklearn.cluster import KMeans
sklearn_pca=PCA(n_components=2)
principalComponents=sklearn_pca.fit_transform(scaled_data)
principalDf=pd.DataFrame(data=principalComponents,columns=['pC1','pC2'])
elbow_method=[]
for i in range(1,11):
  kmeans=KMeans(n_clusters=i,random_state=42)
  kmeans.fit(principalDf)
  elbow_method.append(kmeans.inertia_)
plt.plot(range(1,11),elbow_method,marker='o',linestyle='--')
plt.xlabel('Number of clusters')
plt.ylabel('WCSS')
plt.show()



```

```python
kmeans=KMeans(n_clusters=3,random_state=42)
#kmeans.fit(principalDf)
labels=kmeans.fit_predict(principalDf)
#add the cluster labels to PCA dataframe
kmeans.fit(principalDf)
principalDf['cluster']=labels
principalDf.head()
plt.figure(figsize=(8, 6))
sns.scatterplot(x='pC1',y='pC2',hue='cluster',data=principalDf,palette='Set1')
plt.title('K-Means Clustering')
plt.show()
```

```python
from scipy.cluster.hierarchy import dendrogram,linkage
from sklearn.cluster import AgglomerativeClustering
linked=linkage(principalDf,method='ward')
plt.figure(figsize=(10, 7))
dendrogram(linked,orientation='top',distance_sort='descending',show_leaf_counts=True)
plt.show()
```

```python
from sklearn.metrics import silhouette_score
sil_score = silhouette_score(principalDf, labels)
print(f'Silhouette Score: {sil_score}')
```

```python
#performing DBSCAN
from sklearn.cluster import DBSCAN as DB
import numpy as np
#initilize dbscan
dbscan = DB(eps=0.5, min_samples=5)
dbscan.fit(principalDf)
labels = dbscan.labels_
print(labels)
plt.figure(figsize=(8,6))
plt.scatter(principalDf['pC1'], principalDf['pC2'], c=labels, cmap='viridis')
plt.title('DBSCAN Clustering on PCA Results')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.colorbar(label='Cluster')
plt.show()
```

```python
# Optional: Print the number of clusters (excluding noise points labeled as -1)
n_clusters = len(set(labels)) - (1 if -1 in labels else 0)
print(f'Number of clusters: {n_clusters}')
```
